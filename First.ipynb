```python
import pandas as pd
import os
import logging
import time

from sqlalchemy import create_engine, URL

# Configure logging
logging.basicConfig(
    filename="ingestion_db.log",
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filemode="a"
)

# Connect to MySQL database
connection_url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="Vaishu@23",
    host="localhost",
    port=3306,
    database="inventory"
)

engine = create_engine(connection_url)

# Test the MySQL connection
with engine.connect() as conn:
    print("MySQL connected successfully!")

# Function to ingest DataFrame into MySQL
def ingest_db(df, table_name, engine):
    """Ingest a DataFrame into a MySQL database table."""
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists="replace",
        index=False,
        chunksize=1000
    )

# Function to load CSV files into MySQL
def load_raw_data():
    """Load CSV files and ingest them into MySQL."""
    
    start = time.time()

    data_folder = r"C:\Users\HP\data"

    for file in os.listdir(data_folder):
        if file.lower().endswith(".csv"):

            file_path = os.path.join(data_folder, file)
            table_name = os.path.splitext(file)[0]

            try:
                logging.info(f"Ingesting {file} into MySQL")

                df = pd.read_csv(file_path)

                print(f"Uploading: {table_name}")

                ingest_db(df, table_name, engine)

                logging.info(f"Successfully uploaded {file}")
                print(f"Uploaded: {table_name}")

                del df

            except Exception as e:
                logging.exception(f"Failed to upload {file}: {e}")
                print(f"Error uploading {file}: {e}")

    end = time.time()

    total_time = (end - start) / 60

    logging.info("----------- Ingestion Complete -----------")
    logging.info(f"Total time taken: {total_time:.2f} minutes")

    print(f"\nTotal time taken: {total_time:.2f} minutes")
    print("Ingestion process finished. Check the log file for details.")

# Run the ingestion process
if __name__ == "__main__":
    load_raw_data()




```
